```python
df['country'] = df['country'].str.strip().str.title()
```

```python
df[df['country'] == 'Not Given']['title'].duplicated().sum()
```

```python
df['country'] = df['country'].replace('Not Given', 'Unknown')
```

<h5> i replaced the Not Given feature to Unknown because we can't actually decide the country of the data set because,
    it is a categorical column and we cannot decide the exact country for the movie so we will replace it with Unknown</h5>

```python
temp = df.groupby('title')['director'].unique().reset_index(name='directors')
temp['directors']
```

<h5>Hypothesis 1:<br>
we cannot fill the director by same title? -- No</h5>

```python
possible_fill = temp[temp['directors'].apply(lambda x: ('Not Given' in x) and (len(set(x) - {'Not Given'}) > 0))]
print("Titles that can be safely filled :", len(possible_fill))
```

<h5>The possibility of recovering missing <b>director names</b> using duplicate movie titles was investigated. No matching titles with valid director information were found. Therefore, the missing director values cannot be reliably inferred from the existing dataset.</h5>

```python
df['director'] = df['director'].replace('Not Given', 'Unknown')
```
